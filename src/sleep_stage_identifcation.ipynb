# Reclassification of sleep/ wake stages based on EEG Epochs

```python
import pandas as pd
from pathlib import Path
import re

# change to where data is downloaded
path = Path('/Users/graceburns/Desktop/brainhack/ds003768-download/sourcedata/')

subject_files = sorted(path.glob("sub-*"))

df_all = []

for subject_file in subject_files:
    # Load TSV file into a DataFrame
    df = pd.read_csv(f'{path}/{subject_file.name}', sep='\t')

    df_groups=df.groupby('session')['30-sec_epoch_sleep_stage'].value_counts().unstack(fill_value=0)
    # Display the first few rows

    df_groups['main_stage'] = df_groups.idxmax(axis=1)
    df_groups['subject'] = re.search(r'sub-\d+', subject_file.name).group(0)
    df_all.append(df_groups)

sleep_stage_df = pd.concat(df_all).fillna(0)

sleep_stage_df['sleep'] = sleep_stage_df[['1', '2', '3']].sum(axis=1)

sleep_stage_df.reset_index(inplace=True)
```

```python

idx = sleep_stage_df.groupby('subject')['W'].idxmax()

# Select the full rows based on those indices
result = sleep_stage_df.loc[idx]
#result.reset_index(inplace=True)
result.set_index('subject', inplace=True)

# runs with the most wake stage for each participant (only used if sleep is also the mode stage)
result[['session', 'main_stage', 'W']].query("main_stage == 'W'").to_csv('wake_stage_identification.csv')
```

```python
idx = sleep_stage_df.groupby('subject')['sleep'].idxmax()

# Select the full rows based on those indices
result = sleep_stage_df.loc[idx]
result.reset_index(inplace=True)
result.set_index('subject', inplace=True)

# runs with the most sleep stage for each participant (only used if sleep is NOT the mode stage)
result[['session', 'main_stage', 'sleep']].query("main_stage != 'W'")
```

```python
idx = sleep_stage_df.groupby('subject')['1'].idxmax()

# Select the full rows based on those indices
result = sleep_stage_df.loc[idx]
result.reset_index(inplace=True)
result.set_index('subject', inplace=True)

# runs with the most N1
result[['session', 'main_stage', '1']].query("main_stage == '1'")
```
